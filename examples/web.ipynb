**Purpose:** Create, read, update, and delete web protection objects including URL groups, web filter categories, content filters, scanning exceptions, web policies, user activities, and singleton protection settings.

**Topics:**
- Basic CRUD operations using domain StrEnum objects (`Web`)
- Managing custom URL and domain groups (`Web.URL_GROUPS`)
- Custom web filter categories and productivity classifications (`Web.CATEGORIES`)
- Keyword and MIME-type content filters (`Web.CONTENT_FILTERS`)
- Web traffic scanning exceptions (`Web.EXCEPTIONS`)
- Web filter policies and rule enforcement (`Web.POLICIES`)
- User activity classification (`Web.USER_ACTIVITIES`)
- Singleton proxy inspection settings and user notification banners (`Web.SCAN_SETTINGS`, `Web.USER_NOTIFICATIONS`)
- Filtering and auto-pagination
- Error handling
- Bulk operations

**Requirements:**
- Credentials configured (`.credentials/config.ini`)
- SFOS firewall accessible
- API key with appropriate permissions

#### Setup: Import Libraries and Configure Client

```python
from utils_api import find_credentials_file, print_data

from sfos_api import SFOSClient, Web

# Initialize the SFOS API Client
client = SFOSClient(config_path=find_credentials_file())
print("✅ SFOS API client initialized")
```

#### Create Web Filter URL Groups

Create custom URL and domain groups using the `Web.URL_GROUPS` endpoint object.

URL groups group specific hostnames and domains for use in web policies and exception rules.

API Link: [Create a URL group | Sophos Firewall API](https://docs.sophos.com/nsg/sfos-api/rest-api/api/firewall-config-openapi/create-web-filter-url-group/)

---


```python
# Create a URL group for approved services
endpoint = Web.URL_GROUPS
group_name = "test-web-urlgroup-approved"

payload = {
    "name": group_name,
    "description": "Approved corporate domains and cloud services",
    "urls": [
        "example.com",
        "api.example.com",
        "portal.example.com",
    ],
}

try:
    response = client.post(endpoint, payload)
    print(f"✅ Created URL group '{group_name}'")
    print_data(response)
except Exception as e:
    print(f"Error creating URL group: {e}")
    print_data(str(e))
```

```python
# Create another URL group for blocked domains
endpoint = Web.URL_GROUPS
group_name = "test-web-urlgroup-blocked"

payload = {
    "name": group_name,
    "description": "Blocked advertisement and tracking domains",
    "urls": [
        "ads.example.test",
        "tracker.example.test",
    ],
}

try:
    response = client.post(endpoint, payload)
    print(f"✅ Created URL group '{group_name}'")
    print_data(response)
except Exception as e:
    print(f"Error creating URL group: {e}")
    print_data(str(e))
```

#### Read Web Filter URL Groups

Retrieve URL group information by ID/name or list all URL groups.

API Links: [List URL groups | Sophos Firewall API](https://docs.sophos.com/nsg/sfos-api/rest-api/api/firewall-config-openapi/list-web-filter-url-groups/) | [Get a URL group | Sophos Firewall API](https://docs.sophos.com/nsg/sfos-api/rest-api/api/firewall-config-openapi/get-web-filter-url-group/)

---


```python
# Get a specific URL group by name
endpoint = Web.URL_GROUPS
group_name = "test-web-urlgroup-approved"

try:
    response = client.get(endpoint, id_or_name=group_name)
    print(f"✅ Retrieved URL group '{group_name}'")
    print_data(response)
except Exception as e:
    print(f"Error reading URL group: {e}")
    print_data(str(e))
```

```python
# Get all URL groups
endpoint = Web.URL_GROUPS

try:
    response = client.get(endpoint)

    if "items" in response and response["items"]:
        print(f"✅ Retrieved {len(response['items'])} URL groups")
        print_data(response["items"], include=["name", "description", "urls"])
    else:
        print("No URL groups found")
except Exception as e:
    print(f"Error reading URL groups: {e}")
    print_data(str(e))
```

```python
# Get URL groups with filtering
endpoint = Web.URL_GROUPS
params = {
    "nameContains": "test-web-",
}

try:
    response = client.get(endpoint, params=params)

    if "items" in response:
        print(f"✅ Retrieved {len(response['items'])} URL groups matching filter")
        print_data(response["items"], include=["name", "description", "urls"])
    else:
        print("No URL groups found matching filter")
except Exception as e:
    print(f"Error reading URL groups: {e}")
    print_data(str(e))
```

#### Web Scanning Exceptions

Manage inspection exceptions and scanning bypass lists using `Web.EXCEPTIONS`.

Exceptions allow specific destinations or regex URL patterns to skip HTTPS decryption, certificate validation, or malware scanning.

API Link: [Create a web exception | Sophos Firewall API](https://docs.sophos.com/nsg/sfos-api/rest-api/api/firewall-config-openapi/create-web-filter-exception/)

---


```python
# Create a web scanning exception
endpoint = Web.EXCEPTIONS
exception_name = "test-web-exception-trusted"

payload = {
    "name": exception_name,
    "description": "Skip HTTPS decryption and scanning for trusted update domain",
    "enabled": True,
    "conditions": {
        "urlRegexPatterns": {
            "enabled": True,
            "values": [
                "^([A-Za-z0-9.-]*\\.)?trustedpartner\\.internal/",
            ],
        },
    },
    "skip": {
        "httpsDecryption": True,
        "certificateValidation": False,
        "malwareScanning": True,
        "zeroDayProtection": True,
        "policyCheck": True,
    },
}

try:
    response = client.post(endpoint, payload)
    print(f"✅ Created web exception '{exception_name}'")
    print_data(response)
except Exception as e:
    print(f"Error creating web exception: {e}")
    print_data(str(e))
```

```python
# Read web exceptions
endpoint = Web.EXCEPTIONS

try:
    response = client.get(endpoint)
    if "items" in response and response["items"]:
        print(f"✅ Retrieved {len(response['items'])} web exceptions")
        print_data(response["items"], include=["name", "enabled", "skip"])
    else:
        print("No web exceptions found")
except Exception as e:
    print(f"Error reading web exceptions: {e}")
    print_data(str(e))
```

```python
# Delete a web exception
endpoint = Web.EXCEPTIONS
exception_name = "test-web-exception-trusted"

try:
    response = client.delete(endpoint, id_or_name=exception_name)
    print(f"✅ Deleted web exception '{exception_name}'")
    if response:
        print_data(response)
except Exception as e:
    print(f"Error deleting web exception: {e}")
    print_data(str(e))
```
